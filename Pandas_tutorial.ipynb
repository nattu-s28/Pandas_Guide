## Part 1 — Pandas Series Basics


```python
# Basic Series

print(pd.__version__)       # Current Pandas version

series1 = pd.Series([10, 20, 30, 40]) # Normal Series
series2 = pd.Series([10, 20, 30, 40], name="Numbers") # Customized Name
series3 = pd.Series([10,20,30,40,50], index=["n1","n2","n3","n4","n5"]) # Customized Labels 
series4 = pd.Series([10, 20, 30, 40], dtype=float) # Customized Datatype

print(series1)
print(series2)
print(series3)          # Display the Series
print(series4)
```

## Part 2 — Series


```python
# Creating a Series with custom indexes

s = pd.Series([85, 90, 90, 78, 92, 87, 89, 73], index=["A", "B", "C", "D", "E", "F", "G","H"], name="Scores")

print(s)

print(s.head())         # First 5 Values in Series
print(s.head(6))        # First N Values in Series

print(s.tail())         # Last 5 Values in Series
print(s.tail(6))        # Last N Values in Series

print(s.dtype)          # Returns the Datatype of the Series

print(s.values)         # Returns the Values in the Series

print(s.index)          # Returns the Labels in the Series

print(s.mean())         # Returns the Mean of the Series

print(s.median())       # Returns the Median of the Series

print(s.mode())         # Returns the Mode of the Series

print(s.sum())          # Returns the Sum of the Series

print(s.unique())       # Return Unique Elements from the Series 
```

## Part 6 — Exploring Data


```python
# Exploring the Data

print(students.head())          # Returns First 5 Rows
print(students.head(10))        # Returns First N Rows

print(students.tail())          # Returns Last 5 Rows
print(students.tail(10))        # Returns Last N Rows

print(students.shape)           # Returns the Shape of Rows & Columns

print(students.columns)         # Returns the Name of the Columns

print(students.dtypes)          # Returns the Datatype of the Columns

students.info()                 # Returns the Basic Info of DataFrame

print(students.describe())      # Returns the Statistical Summary of the DataFrame
```

## Part 7 — Selecting Columns


```python
# Selecting Columns

print(students["Name"])                 # Returns one column

print(students[["Name", "Score"]])      # Returns multiple columns

print(students.get("Age"))              # retrieves a column safely 
```

## Part 8 — Selecting Rows


```python
# Selecting Rows

print(students.loc[0])
print(students.loc[0:2, ["Name", "Score"]])  # 'loc[]' selects data using labels & Indices

print(students.iloc[0])
print(students.iloc[0:3, 1:4])               # 'iloc[]' selects data using integer positions

print(students.iloc[0])
print(students.at[0, "Name"])                # 'at[]' accesses one value using labels

print(students.iat[0, 1])                    # 'iat[]' accesses one value using indices
```

## Part 9 — Filtering Data


```python
# Filtering the Data

# Boolean filtering selects rows matching a condition
print(students[students["Age"] > 21])

# '&' applies an AND condition
print(students[(students["Age"] >= 21) & (students["Score"] >= 85)])

# '|' applies an OR condition
print(students[(students["City"] == "Chennai") | (students["Score"] > 90)])

# 'isin()' filters values found in a list
print(students[students["City"].isin(["Chennai", "Delhi"])])

# 'between(') filters values within a range
print(students[students["Age"].between(21, 22)])

# 'query()' filters using an expression
print(students.query("Score >= 90"))
```

## Part 10 — Adding and Modifying Columns


```python
# Modifying the DataFrame

# Create a new column with assignment
students["Bonus"] = 5000

# Create a calculated column
students["Total_Salary"] = students["Salary"] + students["Bonus"]

# np.where() creates values based on a condition
students["Result"] = np.where(students["Score"] >= 80, "Pass", "Improve")

print(students)
```

## Part 11 — Renaming Data


```python
# Renaming the Data

# rename() changes selected column names
renamed = students.rename(columns={"Score": "Exam_Score"})
renamed = students.rename(columns={"Score": "Exam_Score","Bonus" : "Salary_Bonus"})

# 'Copy' the DataFrame for replacing all column names
small_df = students[["ID", "Name", "Age"]].copy()

# 'columns' replaces all column names
small_df.columns = ["Student_ID", "Student_Name", "Student_Age"]

print(renamed.head())
print(small_df.head())
```

## Part 12 — Deleting Data


```python
# Deleting the Data

# 'drop()' removes rows using index labels
without_rows = students.drop(index=[0, 1])

# 'drop()' removes a column
without_bonus = students.drop(columns=["Bonus"])

print(without_rows.head())
print(without_bonus.head())
```

## Part 13 — Missing Values


```python
# Creating sample data containing missing values
missing_df = pd.DataFrame({
    "Name": ["A", "B", "C", "D"],
    "Score": [90, np.nan, 85, np.nan],
    "Age": [21, 22, np.nan, 23]
})

# 'isnull()' detects missing values
print(missing_df.isnull())

# 'isnull().sum()' counts missing values
print(missing_df.isnull().sum())

# 'fillna()' replaces missing values
print(missing_df.fillna(0))

# 'dropna()' removes Rows containing missing values
print(missing_df.dropna())

# 'ffill()' fills using previous values
print(missing_df.ffill())

# 'bfill()' fills using next values
print(missing_df.bfill())
```

## Part 14 — Duplicate Data


```python
# Create sample duplicate data

duplicate_df = pd.DataFrame({
    "Name": ["A", "B", "A", "C"],
    "Score": [90, 80, 90, 85]
})

# 'duplicated()' detects duplicate rows
print(duplicate_df.duplicated())

# 'duplicated().sum()' counts duplicate rows
print(duplicate_df.duplicated().sum())

# 'drop_duplicates()' removes duplicate rows
print(duplicate_df.drop_duplicates())
```

## Part 15 — Sorting Data


```python
# Sorting the DataFrame

# 'sort_values()' sorts using column values
print(students.sort_values("Score"))                    # Assending Order

# ascending=False sorts from highest to lowest
print(students.sort_values("Score", ascending=False))   # Desending Order

# Sort using multiple columns
print(students.sort_values(["City", "Score"]))

# 'sort_index()' sorts using index labels
print(students.sort_index(ascending=False))
```

## Part 16 — Statistical Functions


```python
# Statistical Functions

# 'mean()' returns the average
print(students["Score"].mean())

# 'median()' returns the middle value
print(students["Score"].median())

# 'mode()' returns the most frequent value
print(students["Score"].mode())
print(students["Score"].mode()[0])

# 'min()' returns the smallest value
print(students["Score"].min())

# 'max()' returns the largest value
print(students["Score"].max())

# 'sum()' returns the total
print(students["Score"].sum())

# 'count()' counts non-missing values
print(students["Score"].count())

# 'std()' returns standard deviation
print(students["Score"].std())

# 'var()' returns variance
print(students["Score"].var())

# 'quantile()' returns a selected quantile
print(students["Score"].quantile(0.25))
print(students["Score"].quantile(0.75))
```

## Part 17 — Unique Values


```python
# Unique Values

# 'unique()' returns unique values
print(students["City"].unique())

# 'nunique()' counts unique values
print(students["City"].nunique())

# 'value_counts()' counts each unique value
print(students["City"].value_counts())
```

## Part 18 — GroupBy Operations


```python
# Groupby Operations

# 'groupby()' groups rows using a column
city_groups = students.groupby("City")

# 'mean()' calculates group averages
print(city_groups["Score"].mean())

# 'sum()' calculates group totals
print(city_groups["Salary"].sum())

# 'min()' finds group minimum values
print(city_groups["Score"].min())

# 'max()' finds group maximum values
print(city_groups["Score"].max())

# 'count()' counts values in each group
print(city_groups["Name"].count())

# 'agg()' performs multiple aggregations
print(city_groups["Score"].agg(["mean", "min", "max"]))
```

## Part 19 — Combining DataFrames


```python
# Combing the DataFrames

# Create two DataFrames with a common key
df1 = pd.DataFrame({"ID": [1, 2, 3], "Name": ["Aarav", "Diya", "Rahul"]})
df2 = pd.DataFrame({"ID": [1, 2, 4], "Department": ["AI", "DS", "CS"]})

# concat() combines DataFrames vertically
print(pd.concat([df1, df1], ignore_index=True))

# merge() combines DataFrames using a common column
print(pd.merge(df1, df2, on="ID", how="inner"))

# Left join keeps all rows from the left DataFrame
print(pd.merge(df1, df2, on="ID", how="left"))

# Outer join keeps all rows from both DataFrames
print(pd.merge(df1, df2, on="ID", how="outer"))
```

## Part 20 — String Operations


```python
# String Operations

# str.lower() converts text to lowercase
students["Name_Lower"] = students["Name"].str.lower()

# str.upper() converts text to uppercase
students["Name_Upper"] = students["Name"].str.upper()

# str.len() returns the length of each string
students["Name_Length"] = students["Name"].str.len()

# str.contains() checks for a text pattern
print(students["Name"].str.contains("a", case=False))

# str.startswith() checks starting text
print(students["Name"].str.startswith("A"))

# str.endswith() checks ending text
print(students["City"].str.endswith("i"))

# str.replace() replaces text patterns
print(students["City"].str.replace("Chennai", "CHN", regex=False))

# str.split() splits strings
print(students["Name"].str.split("a"))

print(students.head())
```

## Part 21 — Apply, Map, Replace 

```python
# Other Important Functions

# apply() applies a function to each value
students["Score_Double"] = students["Score"].apply(lambda x: x * 2)

# map() converts categorical values using a mapping
students["Gender_Code"] = students["Gender"].map({"Male": 1, "Female": 0})

# replace() replaces selected values
students["City_Updated"] = students["City"].replace({"Chennai": "CHN"})
```
